# Reproduce the paper figures

Each cell runs one figure script from `src/` on the results committed in `data/csv_Files/`. The
PDFs are written to `figures_final/`. The scripts typeset labels with LaTeX, so a LaTeX
installation is needed.

```python
import subprocess
import sys
from pathlib import Path

REPO = Path.cwd() if (Path.cwd() / "src").is_dir() else Path.cwd().parent


def run(script, *overrides):
    """Run a figure script from the repository root, with optional Hydra overrides."""
    print("$ python", script, *overrides)
    subprocess.run([sys.executable, str(REPO / script), *overrides], cwd=REPO, check=True,
                   stdout=subprocess.DEVNULL)


def show_outputs(*names):
    for name in names:
        path = REPO / "figures_final" / name
        print(("written " if path.exists() else "missing ") + str(path.relative_to(REPO)))
```

## Function evaluations: SCORE vs. differential evolution

```python
run("src/HeatmapPlots.py")
show_outputs("heatmap_comparison.pdf", "heatmap_comparison_vals.pdf")
```

## Data downlinked vs. number of ground stations
